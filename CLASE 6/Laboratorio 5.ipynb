```python
import pandas as pd
import numpy as np
```

```python
# Tabla 1: Base de Empleados (Con errores intencionales)
df_empleados = pd.DataFrame({
"id_emp": ["E001", " E002 ", "e003", "E004"],
"nombre_completo": [" ana perez", "LUIS GOMEZ", "carlos ruiz", "maria diaz "],
"salario_base": ["USD 1200", "USD 1500", "USD 900", "USD 2000"]
})
# Tabla 2: Evaluaciones de Desempeño
df_evaluaciones = pd.DataFrame({
"id_empleado": ["E001", "E002", "E003", "E009"], # Nota: E009 no existe en la base principal
"calificacion": ["A", "B", "A", "C"]
})
```

```python
df_empleados
```

```python
# Limpieza columna id_emp del df_empleados
# df_empleados["id_emp"] = df_empleados["id_emp"].apply(lambda x: x.strip().upper())
df_empleados["id_emp"] = (
    df_empleados["id_emp"]
    .str.strip()
    .str.upper()
)
```

```python
# 2. Limpiar los nombres y convertirlos a formato título
df_empleados["nombre_completo"] = (
    df_empleados["nombre_completo"]
    .str.strip()
    .str.title()
)

# 3. Eliminar "USD " y convertir los salarios a números decimales
df_empleados["salario_base"] = (
    df_empleados["salario_base"]
    .apply(lambda x: float(x.replace("USD ", "").strip())
    )
)
```

```python
df_empleados
```

```python
# Parte 2: Cruce de Tablas (Merge)
df_consolidado = pd.merge(
    df_empleados,
    df_evaluaciones,
    left_on="id_emp", # nombre distinto de llave
    right_on="id_empleado",
    how="left",
    suffixes=("_empleado", "_evaluacion")
)

# Eliminar la columna redundante de la tabla derecha
df_consolidado = df_consolidado.drop("id_empleado", axis=1)
```

```python
df_consolidado
```

```python
#Parte 3. Cálculo condicional de bonos
def calcular_bono(fila):
    # Bono del 20% para calificación A
    if fila["calificacion"] == "A":
        return fila["salario_base"] * 0.20

    # Bono del 10% para calificación B
    elif fila["calificacion"] == "B":
        return fila["salario_base"] * 0.10

    # Bono cero para calificaciones nulas u otras letras
    else:
        return 0.0


# Aplicar la función a cada fila del DataFrame
df_consolidado["bono_asignado"] = df_consolidado.apply(
    calcular_bono,
    axis=1
)

# Calcular el salario neto
df_consolidado["salario_neto"] = (
    df_consolidado["salario_base"]
    + df_consolidado["bono_asignado"]
)

display(df_consolidado)
```
