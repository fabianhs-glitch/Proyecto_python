```python
# CODIGO FINAL CONSOLIDADO
from pathlib import Path
import pandas as pd

datos = Path('/Workspace/Users/fabianhuamansilvestre@gmail.com/CURSO PYTHON/CLASE 6/datasets/ventas_mensuales')

archivos = datos.glob('*') # es un iterable
lista_df=[]

for archivo in archivos:
    if archivo.suffix=='.csv':
        #print(archivo.suffix)
        df=pd.read_csv(archivo,sep=";")
        df['Origen']=archivo.name
        lista_df.append(df)
        #print(df.head(1))
    else:
        print('no es csv')

df_historico=pd.concat(lista_df,ignore_index=True)
print(df_historico.shape)

#df_historico.to_csv('/Workspace/Users/fabianhuamansilvestre@gmail.com/CURSO PYTHON/CLASE 6/datasets/df_historico.csv',index=False)
```
